### Opdracht 3.1h

````python
# h
import random
somrandom = 0
for i in range(20):
    randomgetal = random.randint(1,99)
    somrandom += randomgetal
    print(randomgetal, end=' ')
print("\n")
````

### Opdracht 3.1i

Zet 20 random getallen onder de 100 op het scherm. Bepaal het maximum van de getoonde random getallen.

````python
# i
import random
maxrandom = 0
for i in range(20):
    randomGetal = random.randint(1,99)
    if maxrandom > randomGetal:
        randomGetal == maxrandom
    print(randomGetal, end=" ")
print(randomGetal)
````
